# S8 · Lab 03 — No linealidad: el “interruptor” que hace que las capas sirvan

## Narrativa
Una idea razonable sería:

> “Si un modelo lineal es limitado, apilemos muchos modelos lineales.”

Este laboratorio demuestra una sorpresa:

- **sin activación**, 2 capas lineales son lo mismo que 1 capa lineal,
- con un “interruptor” no lineal (ReLU), la salida **se dobla**.

---

## Qué vas a producir
1) Evidencia numérica: 2 capas lineales ≡ 1 capa lineal.  
2) Evidencia visual: ambas producen **una recta**.  
3) El “interruptor de magia”: añades ReLU y la recta se vuelve **pieza a pieza** (se dobla).


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def relu(z):
    return np.maximum(0, z)
```

```python
# 1) Construimos un ejemplo 1D para visualizar con claridad
x = np.linspace(-3, 3, 400).reshape(-1, 1)

# Dos capas lineales (1 -> 3 -> 1) para que haya "profundidad"
W1 = np.array([[1.2, -0.7, 0.5]])     # (1,3)
b1 = np.array([[0.4, -0.2, 0.0]])     # (1,3)

W2 = np.array([[0.8], [-0.6], [0.3]]) # (3,1)
b2 = np.array([[0.1]])                # (1,1)

# Forward sin activación
Z1 = x @ W1 + b1
y_2lin = Z1 @ W2 + b2

# Capa equivalente
W_eq = W1 @ W2
b_eq = b1 @ W2 + b2
y_1lin = x @ W_eq + b_eq

print("¿2 lineales == 1 lineal?:", np.allclose(y_2lin, y_1lin))
```

```python
# 2) Visualización: sorpresa -> ambas son la MISMA recta
plt.figure()
plt.plot(x, y_2lin, label="2 capas lineales")
plt.plot(x, y_1lin, linestyle="--", label="1 capa equivalente")
plt.title("Sin activación: profundidad = ilusión (sigues teniendo una recta)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
# 3) El interruptor: ReLU entre capas
Z1_relu = relu(Z1)
y_relu = Z1_relu @ W2 + b2

plt.figure()
plt.plot(x, y_2lin, label="Sin activación (recta)")
plt.plot(x, y_relu, label="Con ReLU (se dobla)")
plt.title("Con activación: la función se vuelve no lineal (pieza a pieza)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

## Extra visual: ¿qué hace ReLU “por dentro”?
ReLU actúa como un **filtro**:
- deja pasar valores positivos,
- y anula los negativos a 0.

Eso crea “tramos” (piecewise) y permite que la composición deje de ser una recta global.


```python
# Miramos una neurona oculta para ver el efecto del "corte"
j = 0  # índice de neurona oculta
plt.figure()
plt.plot(x, Z1[:, j], label="Pre-activación (lineal)")
plt.plot(x, relu(Z1[:, j]), label="Post-ReLU (recortado)")
plt.title("ReLU en una neurona: anula negativos, deja positivos")
plt.xlabel("x")
plt.ylabel("valor")
plt.legend()
plt.show()
```
